I first did some small cases and recognized some small recurrence pattern: the sum-player has 2 choices seeing a,b: being a+b or abs(a-b). The sum player here C has to think if he's abs(a-b), if not then he's a+b. if a>=b, F(a,b,a+b) = 2 + F(a,b,a-b) [thinking about being a-b] (A becomes the sum-player, so if a said I don't know after F(a,b,a-b) turn, then C knows that he's a+b and just has to wait his turn: thus +2). The recursion stops once, you come to F(r,r,sum_player)=sum_player+1. Now you could get formulas depending on where the sum player is and a function of abs(a-b), but this problem hides more. If you iterate more, you get some formulas. if a>b, a = qb + r. You can get formulas by iterating last way of thought and getting sum of alterning 1,2 till you get F(a,r,sum_player). Sum_player depends on the parity of q. Actually, once doing that, we understand that the stop point of the algorithm is gcd(a,b). Following our euclidean algorithm like, we'll arrive to F( gcd(a,b), gcd(a,b), ?) and knowing this trivial case, we can stop. This assure that the algorithm will stop. 

```python
def F(a, b, sum=2):
    if a == b:
        return sum + 1
    if sum == 2:
        if b > a:
            q, r = b // a, b % a
            if r == 0:
                q -= 1
            if q % 2 == 0:
                add = (3 * q) // 2
                if r == 0:
                    return add + F(a, a, sum=2)
                return add + F(a, r, sum=2)
            else:
                add = (3 * q - 1) // 2
                if r == 0:
                    return add + F(a, a, sum=1)
                return add + F(a, r, sum=1)
        else:
            q, r = a // b, a % b
            if r == 0:
                q -= 1
            if q % 2 == 0:
                add = (3 * q) // 2
                if r == 0:
                    return add + F(b, b, sum=2)
                return add + F(r, b, sum=2)
            else:
                add = (3 * q + 1) // 2
                if r == 0:
                    return add + F(b,b, sum=0)
                return add + F(r, b, sum=0)
    if sum == 1:
        if b > a:
            q, r = b // a, b % a
            if r == 0:
                q -= 1
            if q % 2 == 0:
                add = (3 * q) // 2
                if r == 0:
                    return add + F(a, a, sum=1)
                return add + F(a, r, sum=1)
            else:
                add = (3 * q + 1)// 2
                if r == 0:
                    return add + F(a, a, sum=2)
                return add + F(a, r, sum=2)
        else:
            q, r = a // b, a % b
            if r == 0:
                q -= 1
            if q % 2 == 0:
                add = (3 * q) // 2
                if r == 0:
                    return add + F(b, b, sum=1)
                return add + F(r, b, sum=1)
            else:
                add = (3 * q - 1) // 2
                if r == 0:
                    return add + F(b, b, sum=0)
                return add + F(b, r, sum=0)
    if sum == 0:
        if b > a:
            q, r = b // a, b % a
            if r == 0:
                q -= 1
            if q % 2 == 0:
                add = (3* q) // 2
                if r == 0:
                    return add + F(a, a, sum=0)
                return add + F(a, r, sum=0)
            else:
                add = (3 * q + 1) // 2
                if r == 0:
                    return add + F(a, a, sum=1)
                return add + F(r, a, sum=1)
        else:
            q, r = a // b, a % b
            if r == 0:
                q -= 1
            if q % 2 == 0:
                add = (3 * q) // 2
                if r==0:
                    return add + F(b, b, sum=0)
                return add + F(r, b, sum=0)
            else:
                add = (3 *q - 1) // 2
                if r==0:
                    return add + F(b, b, sum=2)
                return add + F(r, b, sum=2)
```
